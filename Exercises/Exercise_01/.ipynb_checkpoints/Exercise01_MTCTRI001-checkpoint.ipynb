```python
# Import packages:
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Loading in the dataset (Flynn et al. (1994) batch experiment)
flynn = pd.read_csv("flynn94.csv")

# Check the variable names in the .csv file 
print(flynn.columns)
```

```python
# Exercise Questions to answer in the code 
"""
Plot the data using a software of your choice and answer the following questions:
1. Is the growth exponential?
2. Can you separate the growth curve into some phases?
3. How different is the growth expressed in cellular carbon from the one in cell number and chlorophyll?
4. Is the ammonium consumption approximated by an exponential curve?
5. How would you estimate the parameters of the growth curve?
"""

# The units for the variables were obtained from the table in flynn94.pdf. 
# I used ChatGPT to help convert them into the correct matplotlib form for the axis labels.

# Creating a figure and axis object
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

# Plotting the dataset variables next to each other for better visualization while answering the above questions:
# Cells
axes[0,0].plot(flynn["time"],flynn["cells"], marker="o")
axes[0,0].set_xlabel("Elapsed Time (days)")
axes[0,0].set_ylabel("Number of Cells (10$^6$ mL$^{-1}$)")
axes[0,0].set_title("Growth of $\it{Isochrysis\ galbana}$ cells")

# Chl-a 
axes[0,1].plot(flynn["time"], flynn["chl-a"], marker="o")
axes[0,1].set_xlabel("Elapsed Time (days)")
axes[0,1].set_ylabel("Chlorophyll-a (ng mL$^{-1}$)")
axes[0,1].set_title("Chlorophyll-a")

# CellC
axes[1,0].plot(flynn["time"], flynn["cellC"], marker="o")
axes[1,0].set_xlabel("Elapsed Time (days)")
axes[1,0].set_ylabel("Cellular Carbon (µg C mL$^{-1}$)")
axes[1,0].set_title("Cell Carbon")

# Ammonium
axes[1,1].plot(flynn["time"], flynn["ammonium"], marker="o")
axes[1,1].set_xlabel("Elapsed Time (days)")
axes[1,1].set_ylabel("Ammonium (µg N mL$^{-1}$)")
axes[1,1].set_title("Ammonium drawdown")

# Need to add tick markers for plot interpretation of the plots. 
for ax in axes.flat:
    ax.set_xticks(np.arange(0, 26, 5)) # tick marks and labels every 5 days
    ax.set_xticks(np.arange(0, 26, 1), minor=True) # tick marks every other day (without labels)
    ax.grid(which="major", alpha=0.6)
    ax.grid(which="minor", alpha=0.2)

# Adjusting the subplots spacing so they fit better.
fig.tight_layout()

# Showing the figures
plt.show()
```

## 1. Is the growth exponential?

The growth is exponential. Initially, the cell concentration experiences a lag phase, followed by an exponential increase between Days 6 and 18. Afterward, it stabilizes at about 2.0 × 10^6 cells/mL, reaching a plateau.

## 2. Can you separate the growth curve into some phases?

**Lag phase:** (Days 0-6) Cell concentration is low and grows slowly.

**Exponential growth phase:** (Days 6-18) Cell numbers increase quickly.

**Plateau phase:** (Days 18-25) Cell concentration plateaus and remains constant. During this phase, ammonium is depleted, and growth is limited by nutrients.

## 3. How different is the growth expressed in cellular carbon from the one in cell number and chlorophyll?

Cellular carbon growth differs from cell count and chlorophyll-a. During the exponential phase, all three variables increase rapidly. Cell concentration plateaus around Day 18, indicating slowed population growth. Meanwhile, cellular carbon continues to increase beyond Day 18, indicating ongoing carbon accumulation despite reduced cell division. Chlorophyll-a increases until around Days 16-17, then declines during the plateau phase (carrying capacity), likely due to nutrient limitation.

## 4. Is the ammonium consumption approximated by an exponential curve?

The ammonium drawdown does not follow an exponential curve. Instead, ammonium levels decrease slowly in the first few days, then decline rapidly between Days 7 and 12. After Day 12, ammonium is completely depleted and remains at zero for the remainder of the batch experiment. This indicates that ammonium drawdown is not an exponential decline; it is governed by phytoplankton growth and nutrient uptake.

## 5. How would you estimate the parameters of the growth curve?

The parameters of the growth curve can be determined from the various phases of *Isochrysis galbana* cell growth (as described in Question 2 above). The initial population size is derived from the first data point. The growth rate is estimated during the exponential phase (around Days 6-18) by measuring how rapidly the cell concentration increases with time. The carrying capacity is obtained from the plateau, i.e., when the *Isochrysis galbana* reaches carrying capacity.
