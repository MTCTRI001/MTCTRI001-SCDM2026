## Question 1
Modify the Euler-forward method implemented in the codes at the beginning of this section (either the python or matlab version) and change it into the Euler-backward method.

```python
# import packages 
import pylab as pl
import matplotlib.pyplot as plt
```

```python
# Forward Euler Method 
# This is a numerical solution of the decay equation where:
t0 = 0 # initial time (days)
tn = 20 # end time (days)
dt = 10 # time interval (days)
k = 0.05 # decay rate (d^-1)
C0 = 0.5 # mg chl-a m^-3 (prey)

def f(k,C): # define the function for the RHS of the ODE: dC/dt = -kC
    
    return -k*C
```

```python
# Euler-forward method

# Calc the number of timesteps needed to run the model
Ntot = (pl.floor((tn-t0)/dt)+1).astype(int)

# Create empty arrays
C = pl.zeros(Ntot) # array used to store chlorophyll-a conc
t = C.copy() # array used to store time

C[0] = C0 # initial conc
t[0] = t0 # initial time

# Start the loop to calculate conc over time
for n in range(Ntot-1):
    
    C[n+1] = C[n] + f(k, C[n])*dt # Euler-forward update
    t[n+1] = t[n] + dt # update the time

# Plotting the Euler-forward solution
plt.plot(t, C)
plt.xlabel("Time (days)")
plt.ylabel("Chlorophyll-a Concentration $(mg\ chl\ m^{-3})$")
plt.title("Chl-a Decay Using the Euler-forward Method")

# Showing the figure
plt.show()
```

```python
# Euler-backward method

dt = 10 # time interval (days)

# Calc the number of timesteps needed to run the model
Ntot = (pl.floor((tn-t0)/dt)+1).astype(int)

# Create empty arrays
C = pl.zeros(Ntot) # array used to store chlorophyll-a conc
t = C.copy() # array used to store time

C[0] = C0 # initial conc
t[0] = t0 # initial time

# Start the loop to calculate conc over time
for n in range(Ntot-1):
    
    C[n+1] = C[n] / (1 + k*dt) # Euler-backward update
    t[n+1] = t[n] + dt # update the time

# Plot the Euler-backward solution
plt.plot(t, C)
plt.xlabel("Time (days)")
plt.ylabel("Chlorophyll-a Concentration $(mg\ chl\ m^{-3})$")
plt.title("Chl-a Decay Using the Euler-backward Method")

# Showing the figure
plt.show()
```

## Question 2
Compare the results for different values of ∆t and upload your code and results on the GitHub repository

```python
# different values of ∆t

# Define a func for the Euler-backward method
def euler_backward(dt):
    
    # Calc the number of timesteps needed to run the model
    Ntot = int(pl.floor((tn-t0)/dt)+1)
    
    # Create empty arrays
    C = pl.zeros(Ntot) # array used to store chlorophyll-a conc
    t = C.copy() # array used to store time
    
    # initial conditions
    C[0] = C0 # initial conc
    t[0] = t0 # initial time
    
    # Start the loop to calculate conc over time
    for n in range(Ntot-1):
        
        C[n+1] = C[n] / (1 + k*dt) # Euler-backward update
        t[n+1] = t[n] + dt # update the time
        
    return t, C # return the time and concentration arrays

# Timestep values to be plotted 
dt_values = [10, 5, 2, 1]

# Create the figure
plt.figure(figsize=(8,6))

# Plot the Euler-backward solution for each timestep
for dt in dt_values:
    
    t, C = euler_backward(dt)
    plt.plot(t, C, label=f"$\Delta t$ = {dt} d")

# Add labels and legend
plt.xlabel("Time (days)")
plt.ylabel("Chlorophyll-a Concentration $(mg\ chl\ m^{-3})$")
plt.title("Euler-backward Method at Different Timestep Values")
plt.legend()

# Showing the figure
plt.show()
```

The decay ODE was evaluated using the Euler-backward method with four different time steps (Δt = 1, 2, 5, & 10 days). All runs produced smooth, decaying curves without any negative values, confirming the scheme's unconditional stability regardless of the time step. However, larger time steps produced less accurate solutions, with the decay occurring more slowly than expected, leading to an overestimation of the Chl-a concentration. On the other hand, smaller time steps produced results closer to the true solution. This indicates that although the scheme remains stable regardless of the time step, accuracy depends on the selected Δt value.
